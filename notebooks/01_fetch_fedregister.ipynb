## Setup

```python
import re, json, requests, pandas as pd, html
from bs4 import BeautifulSoup
from pathlib import Path
from pprint import pprint
from pathlib import Path
```

## Variables 

```python
URL = "https://www.federalregister.gov/api/v1/documents.json"

# Directory to save raw data
output_dir = Path("../data/raw/fedregister")
output_dir.mkdir(parents=True, exist_ok=True)

# File paths for saving data
json_path = output_dir / "fedregister_raw.json"
parquet_path = output_dir / "fedregister_raw.parquet"
```

## Main

```python
# Search terms and years
terms = ["tariff", "section 301", "section 232", "safeguard"]
years = [2020, 2021, 2022, 2023, 2024]
```

```python
params = {
    "conditions[term]": term,
    "conditions[publication_date][gte]": f"{year}-01-01",
    "conditions[publication_date][lte]": f"{year}-12-31",
    "per_page": 1000,
    "page": page,
    "order": "newest",
}

# Add repeated keys separately
params_list = list(params.items()) + [
    ("conditions[agencies][]", 2),
    ("conditions[agencies][]", 54),
    ("conditions[agencies][]", 491),
]
params_list
```

```python
requests.get(
    URL,
    params={
        "conditions[term]": term,
        "conditions[publication_date][gte]": f"{year}-01-01",
        "conditions[publication_date][lte]": f"{year}-12-31",
        "per_page": 1000,
        "page": page,
        "order": "newest",
        "conditions[agencies]": [
            "commerce-department",
        ],
    },
).json()
```

```python
agencies = requests.get(
    "https://www.federalregister.gov/api/v1/agencies"
).json()
for agency in agencies:
    if agency["id"] in [2, 54, 491]:  # White House, Commerce Department, USTR
        print(agency["slug"])
```

```python
rows = []  # List to store all fetched documents
seen_ids = set()  # Set to track unique document IDs

for term in terms:
    for year in years:
        print(f"\n{'='*40} Term: '{term}', Year: {year} {'='*40}")
        for page in range(1, 1000):
            print(f"\n{'-'*20} Page: {page} {'-'*20}")
            params = {
                "conditions[term]": term,
                "conditions[publication_date][gte]": f"{year}-01-01",
                "conditions[publication_date][lte]": f"{year}-12-31",
                "per_page": 1000,
                "page": page,
                "order": "newest",
                "conditions[agencies]": [
                    2,
                    54,
                    491,
                ],  # White House, Commerce, USTR
            }

            response = requests.get(URL, params=params)
            if response.status_code != 200:
                break

            results = response.json().get("results", [])
            if not results:
                break

            for doc in results:
                doc_id = doc["document_number"]
                is_duplicate = doc_id in seen_ids
                print(
                    f"{doc_id} | {doc['publication_date']}"
                    + (" | DUPLICATE" if is_duplicate else "")
                )
                seen_ids.add(doc_id)
                rows.append(doc)

            if len(results) < 1000:
                break

print(f"\nTotal raw documents collected: {len(rows)}")
```

```python
# Collected data to DataFrame
df = pd.DataFrame(rows)
print(f"\nTotal raw documents collected: {len(df)}")

# DataFrame to JSON and Parquet formats
df.to_json(json_path, orient="records", lines=True)
df.to_parquet(parquet_path, engine="pyarrow", index=False)
print(f"Data saved to {json_path} and {parquet_path}")
```

### Test Query - pages

The Federal Register API returns results in pages, each containing up to 1000 documents.
If a request returns exactly 1000 items, more results may be available.
By default, only the first page is retrieved.
To access all documents, iterate through pages using the `page` parameter until no more results are returned.

```python
terms = ["tariff"]  # , "section 301", "section 232", "safeguard"]

rows = []

for term in terms:
    print(f"\nTerm: '{term+"' ":=<120}")

    page = 1  # start at page 1 (default, but I am explicit here)

    while True:
        print(f"\nPage: {f"{page} ":-<120}")
        # request one page at a time, up to 1000 rows
        params = {
            "conditions[term]": term,
            "per_page": 1000,  # default is 20
            "page": page,
            "order": "newest",  # default
        }

        response = requests.get(URL, params=params)

        if response.status_code == 200:
            results = response.json()["results"]
            if (
                not results
            ):  # results empty - means that there are no more pages
                break
            rows.extend(results)
            if (
                len(results) < 1000
            ):  # less than 1000 rows - means that this is the last page
                break
            page += 1
            # print("Tariff results:", 120 * "-", "\n")
            # for doc in data["results"]:
            # pprint(doc["title"])
            # print(120 * ".", "\n")
            # print(120 * "-", "\n")
            print(f"Number of documents: {len(results)}")
```

### Test Query - date slicing instead of pagination

#### Single year, single page

```python
params_date_slicing_single_year = {
    "conditions[term]": "tariff",
    "conditions[publication_date][gte]": "2020-01-01",
    "conditions[publication_date][lte]": "2020-12-31",
    "per_page": 1000,
    "order": "newest",
}

response_date_slicing_single_year = requests.get(
    URL, params=params_date_slicing_single_year
)

if response_date_slicing_single_year.status_code == 200:
    data_date_slicing_single_year = response_date_slicing_single_year.json()
    seen_ids = set()
    for doc in data_date_slicing_single_year["results"]:
        doc_id = doc["document_number"]
        dup = " | DUPLICATE" if doc_id in seen_ids else ""
        print(f"{doc_id} | {doc['publication_date']}{dup}")
        seen_ids.add(doc_id)
    print(f"\nTotal unique documents: {len(seen_ids)}")
```

#### Single year, all pages

```python
seen_ids = set()
page = 1

while True:
    params_filtered_multiple = {
        "conditions[term]": "tariff",
        "conditions[publication_date][gte]": "2020-01-01",
        "conditions[publication_date][lte]": "2020-12-31",
        "per_page": 1000,
        "page": page,
        "order": "newest",
    }

    response = requests.get(URL, params=params_filtered_multiple)
    if response.status_code != 200:
        break

    results = response.json().get("results", [])
    print(f"Number of documents: {len(results)}")

    if not results:
        break

    for doc in results:
        doc_id = doc["document_number"]
        dup = " | DUPLICATE" if doc_id in seen_ids else ""
        print(f"{doc_id} | {doc['publication_date']}{dup}")
        seen_ids.add(doc_id)

    if len(results) < 1000:
        break

    page += 1

print(f"\nTotal unique documents: {len(seen_ids)}")
```

That worked well, single year slice returned non-duplicated documents without going into infinite loop like before
